```python
# Filtrar solo las filas donde DIAGNOSIS es 'PD' o 'CONTROL'
df_filtrado = df_unido[df_unido['DIAGNOSIS'].isin(['PD', 'Control'])].copy()

# Install scikit-learn if it's not already installed
!pip install scikit-learn

# Codificar la variable diagnosis (pd=1, control=0)
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

le = LabelEncoder()
df_filtrado['diagnosis_encoded'] = le.fit_transform(df_filtrado['DIAGNOSIS'])

# Verificar la codificación
print("Codificación:")
print(f"PD = {le.transform(['PD'])[0]}")
print(f"Control = {le.transform(['Control'])[0]}")
print("\nDistribución:")
print(df_filtrado['DIAGNOSIS'].value_counts())

# Preparar features y target
X = df_filtrado.drop(columns=['DIAGNOSIS', 'diagnosis_encoded'])
y = df_filtrado['diagnosis_encoded']

# Dividir en train y test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"\nTamaño del dataset filtrado: {df_filtrado.shape}")
print(f"X_train: {X_train.shape}, X_test: {X_test.shape}")
```

```python
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Create imputador (rellena NaN con la media de cada columna)
imputer = SimpleImputer(strategy='mean')

# Select only numeric columns for imputation and modeling
X_train_numeric = X_train.select_dtypes(include=np.number)
X_test_numeric = X_test.select_dtypes(include=np.number)


# Ajustar e imputar en el conjunto de entrenamiento
X_train_imputed = imputer.fit_transform(X_train_numeric)

# Aplicar el mismo imputador al conjunto de prueba
X_test_imputed = imputer.transform(X_test_numeric)

# Entrenar el modelo con los datos sin NaN
model = LogisticRegression(max_iter=1000)
model.fit(X_train_imputed, y_train)

# Predicción
y_pred = model.predict(X_test_imputed)

# Evaluar
print("Exactitud:", accuracy_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, confusion_matrix, ConfusionMatrixDisplay, roc_curve, auc

# Select only numeric columns for the model
X_train_numeric = X_train.select_dtypes(include=np.number)
X_test_numeric = X_test.select_dtypes(include=np.number)

rf = RandomForestClassifier(random_state=42)
rf.fit(X_train_numeric, y_train)

y_pred_rf = rf.predict(X_test_numeric)

print(f"Exactitud Random Forest: {accuracy_score(y_test, y_pred_rf):.2f}")

auc_rf = roc_auc_score(y_test, y_pred_rf)
print(f"AUC Random Forest: {auc_rf:.2f}")


cm = confusion_matrix(y_test, y_pred_rf)

ConfusionMatrixDisplay(cm).plot(cmap="Blues")
plt.show()


y_proba_rf = rf.predict_proba(X_test_numeric)[:, 1]

# Calculate ROC curve for the positive class
fpr, tpr, _ = roc_curve(y_test, y_proba_rf)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(8,6))
plt.plot(fpr, tpr, label=f"Random Forest (AUC = {roc_auc:.3f})")

# Línea base
plt.plot([0,1], [0,1], 'k--')

print(f"AUC Random Forest: {roc_auc:.2f}")

plt.xlabel("Tasa de Falsos Positivos")
plt.ylabel("Tasa de Verdaderos Positivos")
plt.title("Curva ROC para Clasificación Binaria (PD vs Control)")
plt.xlabel("Tasa de Falsos Positivos")
plt.ylabel("Tasa de Verdaderos Positivos")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# ============================================
# 🔥 OPTIMIZAR EL UMBRAL PARA MEJORAR CONTROL
# ============================================

import numpy as np
from sklearn.metrics import confusion_matrix, classification_report

# 1. Probabilidades originales
y_proba = best_rf.predict_proba(X_test_numeric)[:, 1]

# 2. Vamos a buscar el UMBRAL óptimo
umbrales = np.arange(0.10, 0.90, 0.01)

mejor_umbral = 0.50
mejor_balance = -999

mejor_cm = None

for t in umbrales:
    y_pred_temp = (y_proba >= t).astype(int)
    
    cm = confusion_matrix(y_test, y_pred_temp)
    tn, fp, fn, tp = cm.ravel()

    # Balance para mejorar CONTROL y no destruir PD
    score = (tn / (tn + fp)) + (tp / (tp + fn))

    if score > mejor_balance:
        mejor_balance = score
        mejor_umbral = t
        mejor_cm = cm

print(f"\n🔎 Mejor umbral encontrado: {mejor_umbral:.2f}")

# 3. Predicción final con el mejor umbral
y_pred_opt = (y_proba >= mejor_umbral).astype(int)

# ============================================
# 4. NUEVO REPORT
# ============================================

print("\n🔵 Classification report optimizado:\n")
print(classification_report(y_test, y_pred_opt, target_names=le.classes_))

# ============================================
# 5. NUEVA MATRIZ DE CONFUSIÓN
# ============================================

plt.figure(figsize=(6,4))
sns.heatmap(mejor_cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=le.classes_,
            yticklabels=le.classes_)
plt.title(f"Matriz de confusión optimizada (Umbral={mejor_umbral:.2f})")
plt.xlabel("Predicción")
plt.ylabel("Verdadero")
plt.show()

# ============================================
# 6. AUC SE MANTIENE
# ============================================

auc = roc_auc_score(y_test, y_proba)
print(f"\nAUC del modelo (NO cambia): {auc:.4f}")
```

```python
# ============================================
# 0. LIBRERÍAS
# ============================================
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report, confusion_matrix,
    roc_auc_score, roc_curve, accuracy_score
)
from sklearn.ensemble import RandomForestClassifier
from sklearn.calibration import CalibratedClassifierCV
from imblearn.over_sampling import SMOTE
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================
# 1. FILTRAR Y PREPARAR DATOS
# ============================================
df_filtrado = df_unido[df_unido['DIAGNOSIS'].isin(['PD', 'Control'])].copy()

le = LabelEncoder()
df_filtrado["diagnosis_encoded"] = le.fit_transform(df_filtrado["DIAGNOSIS"])

X = df_filtrado.drop(columns=["DIAGNOSIS", "diagnosis_encoded"])
y = df_filtrado["diagnosis_encoded"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

X_train_numeric = X_train.select_dtypes(include=np.number)
X_test_numeric = X_test.select_dtypes(include=np.number)

print("Datos preparados.")


# ============================================
# 2. RANDOM FOREST INICIAL PARA IMPORTANCIAS
# ============================================
rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    max_features="sqrt",
    class_weight="balanced",
    n_jobs=-1,
    random_state=53
)
rf.fit(X_train_numeric, y_train)

importancias = rf.feature_importances_
df_imp = pd.DataFrame({
    "Variable": X_train_numeric.columns,
    "Importancia": importancias
}).sort_values(by="Importancia", ascending=False)

# Filtrar variables > 0.00009
df_imp_filtrado = df_imp[df_imp["Importancia"] > 0.00009]
vars_finales = df_imp_filtrado["Variable"].tolist()

X_train_filtrado = X_train_numeric[vars_finales]
X_test_filtrado = X_test_numeric[vars_finales]

print(f"Variables finales: {len(vars_finales)}")


# ============================================
# 3. BALANCEO CON SMOTE
# ============================================
sm = SMOTE(random_state=42)
X_train_bal, y_train_bal = sm.fit_resample(X_train_filtrado, y_train)

print("Balanceo aplicado:")
print("Antes:", np.bincount(y_train))
print("Después:", np.bincount(y_train_bal))


# ============================================
# 4. RANDOM FOREST MEJORADO
# ============================================
rf_final = RandomForestClassifier(
    n_estimators=500,
    max_depth=None,
    min_samples_split=5,
    min_samples_leaf=2,
    max_features="sqrt",
    class_weight="balanced_subsample",
    n_jobs=-1,
    random_state=42
)
rf_final.fit(X_train_bal, y_train_bal)

print("Modelo Random Forest entrenado.")


# ============================================
# 5. CALIBRACIÓN DEL MODELO
# ============================================
calibrated_rf = CalibratedClassifierCV(rf_final, method="sigmoid")
calibrated_rf.fit(X_train_bal, y_train_bal)

print("Modelo calibrado.")


# ============================================
# 6. PREDICCIONES + AJUSTE DE UMBRAL
# ============================================
y_proba = calibrated_rf.predict_proba(X_test_filtrado)[:, 1]

# UMBRAL OPTIMIZADO (puedes moverlo entre 0.30 y 0.45)
threshold = 0.35
y_pred_opt = (y_proba >= threshold).astype(int)


# ============================================
# 7. MÉTRICAS
# ============================================
auc = roc_auc_score(y_test, y_proba)
print(f"\nAUC: {auc:.4f}")
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_opt, target_names=le.classes_))


# ============================================
# 8. MATRIZ DE CONFUSIÓN
# ============================================
cm = confusion_matrix(y_test, y_pred_opt)

plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=le.classes_,
            yticklabels=le.classes_)
plt.title(f"Matriz de confusión (Threshold={threshold})")
plt.xlabel("Predicción")
plt.ylabel("Real")
plt.show()


# ============================================
# 9. CURVA ROC
# ============================================
fpr, tpr, _ = roc_curve(y_test, y_proba)

plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, label=f"AUC = {auc:.4f}")
plt.plot([0,1], [0,1], "--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Curva ROC - Random Forest Mejorado")
plt.legend()
plt.show()
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score

# ================================================================
# 1. Crear nuevos X_train y X_test con solo variables > 0.00009
# ================================================================
X_train_filtrado = X_train_numeric[vars_finales]
X_test_filtrado = X_test_numeric[vars_finales]

print("\nDimensión final de X_train filtrado:")
print(X_train_filtrado.shape)

# ================================================================
# 2. Calcular el peso para datos desbalanceados
# ================================================================
import numpy as np

ratio = y_train.value_counts()[0] / y_train.value_counts()[1]
print("scale_pos_weight recomendado:", ratio)

# ================================================================
# 3. Entrenar XGBoost para datos NO balanceados
# ================================================================
xgb_final = XGBClassifier(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.7,
    colsample_bytree=0.7,
    
    # ⚡ clave para datos desbalanceados
    scale_pos_weight=ratio,

    eval_metric="logloss",
    random_state=48,
    n_jobs=-1
)

xgb_final.fit(X_train_filtrado, y_train)

print("\nModelo final XGBoost entrenado SOLO con variables importantes (Importancia >0.00009).")

# ================================================================
# 4. Calcular AUC del modelo final
# ================================================================
y_pred_proba = xgb_final.predict_proba(X_test_filtrado)[:, 1]

auc = roc_auc_score(y_test, y_pred_proba)
print(f"\nAUC del modelo final (XGBoost balanceado): {auc:.4f}")
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score, precision_score, recall_score, f1_score

# ================================================================
# 5. Predicciones finales
# ================================================================
y_pred = xgb_final.predict(X_test_filtrado)

# ================================================================
# 6. Matriz de confusión
# ================================================================
matriz_conf = confusion_matrix(y_test, y_pred)
print("\nMatriz de Confusión:")
print(matriz_conf)

# ================================================================
# 7. Métricas del modelo
# ================================================================
acc  = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec  = recall_score(y_test, y_pred)
f1   = f1_score(y_test, y_pred)

print("\nMétricas del modelo XGBoost:")
print(f"Accuracy:  {acc:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall:    {rec:.4f}")
print(f"F1-score:  {f1:.4f}")

# ================================================================
# 8. Reporte de clasificación
# ================================================================
print("\nReporte de clasificación:")
print(classification_report(y_test, y_pred))
```
